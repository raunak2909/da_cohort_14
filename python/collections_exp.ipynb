```python
# Find the total count of elements/items from the below list 
# whose length is greater than 5 and the first and last char are same
a = ["organic", "pop", "clinic", "morning", "gathering"]

count = 0
for name in a:
    if len(name)>5 and (name[0]==name[-1]):
        count+=1

print(count)
```

```python
# create 2 seperate list for odd and even
a = [2,5,4,2,8,7,6,5,11,31,34,6,7,88,77,85]
odd = []
even = []

for no in a:
    if no%2==0:
        even.append(no)
    else:
        odd.append(no)

print(even, odd)        
```

```python
# write a program to remove duplicates from a given list of no 

no = [5,4,32,2,7,6,5,7,6,3,9,8,3,5,6,2,1,6,9,11,34,55,2,4,3,2]

uniqueNo = list(set(no))

uniqueNo
```

```python
# write a program to remove duplicates from a given list of no w/o converting it into set
no = [5,4,32,2,7,6,5,7,6,3,9,8,3,5,6,2,1,6,9,11,34,55,2,4,3,2]
uniqueNo = []


for n in no:
    if n not in uniqueNo:
        uniqueNo.append(n)
        
print(uniqueNo)
```
