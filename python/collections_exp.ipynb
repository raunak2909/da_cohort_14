```python
# Find the total count of elements/items from the below list 
# whose length is greater than 5 and the first and last char are same
a = ["organic", "pop", "clinic", "morning", "gathering"]

count = 0
for name in a:
    if len(name)>5 and (name[0]==name[-1]):
        count+=1

print(count)
```

```python
# create 2 seperate list for odd and even
a = [2,5,4,2,8,7,6,5,11,31,34,6,7,88,77,85]
odd = []
even = []

for no in a:
    if no%2==0:
        even.append(no)
    else:
        odd.append(no)

print(even, odd)        
```

```python
# Dictionary
# - unordered collection with pair of key and value
# - surrounded by {}
# - keys must be unique
# - mutable

stu = ["Raghav", 16, 77.43, True, "raghav@gmail.com"]

stuDetail = {
    "name" : "Raman",
    "age" : 16,
    "percentage" : 77.43,
    "isFeePaid" : True,
    "mail" : "raman@gmail.com",
    "age" : 19
}
print(stuDetail)
```

```python
print(stuDetail["age"])
```

```python
# add
stuDetail["city"] = "Jodhpur"
print(stuDetail)
```

```python
# update
stuDetail["percentage"] = 89.43
print(stuDetail)
```

```python
# delete
del stuDetail["mail"]
```

```python
print(stuDetail)
```

```python
# get all the keys
for a in stuDetail.keys():
    print(a)
```

```python
# get all the values
for a in stuDetail.values():
    print(a)
```

```python
# get all the items
for a,b in stuDetail.items():
    print(a, ":", b)
```
