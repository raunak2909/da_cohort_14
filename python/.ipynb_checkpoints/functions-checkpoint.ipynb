```python
def add(n1,n2): #params
    print(f"The sum of {n1} and {n2} is {n1+n2}")

add(11,21) #args    



add(100,200)



add(500, 200)



add(77, 32)
```

```python
def add(n1,n2,n3=0,n4=0): #params
    print(f"The sum is {n1+n2+n3+n4}")


add(3,4,2,5)
add(3,7)    
```

```python
# return statement

def add(a,b):
    print(f"The sum is {a+b}")

def addition(n1,n2): #explicit return
    return n1+n2


n = addition(5,6) 
print(n**2)
```

```python
def ageChecker(age):
    if age>=18:
        print("You're eligible")
        return True
    else: 
        print("You're not eligible")
        return False

a = int(input("Enter the age: "))

# these instructions to be executed when eligible

# these instructions to be executed when not eligible

if ageChecker(a):
    print("These are the Tasks for Adults!!")

else:
    print("These are the Tasks for Kids!!")
```
