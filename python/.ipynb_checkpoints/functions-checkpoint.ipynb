```python
def ageChecker(age):
    if age>=18:
        print("You're eligible")
        return True
    else: 
        print("You're not eligible")
        return False

a = int(input("Enter the age: "))

# these instructions to be executed when eligible

# these instructions to be executed when not eligible

if ageChecker(a):
    print("These are the Tasks for Adults!!")

else:
    print("These are the Tasks for Kids!!")
```

```python
# factorial : 5! = 5*4*3*2*1 = 120
sum = 1
for n in range(1,6):
    sum *= n

print(sum)
```

```python
# 5*4*3*2*1
def fact(n):
    if n == 1: #terminating cond
        return 1
    return n * fact(n-1) 

# 5 * 4 * 3 * 2 * 1
fact(5)
# fibonacci series : 0,1,1,2,3,5,8...
```

```python
#lambda functions

add = lambda n1,n2 : n1+n2
prod = lambda n1,n2 : n1*n2
sqr = lambda a : a*a
calcMarks = lambda h,e,m,s,ss : h+e+m+s+ss

print(add(5,6))
print(prod(3,4))
print(sqr(7))
print(calcMarks(90,89,67,78,89))
```

```python
def prod(*args): #(34,23,5,6,78)
    sum = 1
    t = args
    for n in t:
        sum *= n
    print(sum)    

prod(4)
prod(2,3,4,5)
prod(11,23)
prod(34,23,5,6,78)
```
