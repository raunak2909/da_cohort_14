```python
# math
import math

print(math.pow(25,0.5)) # **
print(round(78.45))
print(math.ceil(78.45))
print(math.floor(78.75))# //
print(math.trunc(78.75))# //
print(math.factorial(5))
print(math.sqrt(25))
print(math.pi)
print(math.fmod(56,2))
```

```python
# datetime
import datetime as dt

currDate = dt.datetime.now()
print(currDate.date())
print(currDate.day)
print(currDate.year)
print(currDate.month)
print(currDate.hour)
print(currDate.minute)


date = dt.date(2026,11,30)
print(date.weekday())
```

```python
my_date = "2026/12/01"
print(my_date)
date = dt.datetime.strptime(my_date, "%Y/%m/%d")
print(date)
print(date.year)
print(date.month)
print(date.weekday())
```

```python
new_date = dt.datetime.strftime(date, "%a, %d %B %Y")
print(new_date)
```
