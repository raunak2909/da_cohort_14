```python
# while - when you want some code to iterate till the condition is true - condition specific


for i in range(10, 20, 1): 
    print(f"{i}.Hello")

n = 10
while n<20:
    print(f"{n}.Hello")
    n += 1
    
```

```python
choice = "Yes" 

while choice in ("YES", "Yes", "yes", "Y", "y"):
    n1 = int(input("Enter no1: "))
    n2 = int(input("Enter no2: "))
    sum = n1+n2
    print(f"The sum of {n1} and {n2} is {sum}")
    choice = input("Do you want to add more no(Yes/No)?")
```

```python
# while True

while True:
    n1 = int(input("Enter no1: "))
    n2 = int(input("Enter no2: "))
    sum = n1+n2
    print(f"The sum of {n1} and {n2} is {sum}")
    choice = input("Do you want to add more no(Yes/No)?")
    if choice in ("No", "no", "NO", "N", "n"):
        break
```

```python
sum = 0

while True:
    n = int(input("Enter any no to add:"))
    sum += n
    choice = input("Do you want to add more no(Yes/No)?")
    if choice in ("No", "no", "NO", "N", "n"):
        break
print(f"The sum till now is {sum}")        
```

```python
# billing system
while True:
    name = input("Enter your name: ")
    while True:
        mobNo = input("Enter your mobile no in 10 digits:")
        if len(mobNo)==10:
            break
        else:
            print("Enter a valid 10 DIGIT no")
    
    amount = 0
    while True:
        price = float(input("Enter price of product: "))
        qty = int(input("Enter qty: "))
        
        amount += (price * qty)
        add_more = input("Do you have more items in cart(Yes/No)?")
        if add_more not in ("Yes", "yes", "YES", "Y", "y"):
            break
    print("-"*50)
    print(f"Name : {name}")
    print(f"MobNo : {mobNo}")
    print("-"*50)
    print(f"Total bill : {amount}") 
    print("-"*50)
    
    more_cust = input("Do we have more customers in queue(Yes/No)?")
    if more_cust not in ("Yes", "yes", "YES", "Y", "y"):
        break
```
