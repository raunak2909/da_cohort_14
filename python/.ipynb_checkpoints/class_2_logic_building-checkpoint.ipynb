```python
productName = "SmartPhone"
price = 20000
discount = "10"
cost = 15000
qty = 10

totalSales = price * qty
totalDiscount = totalSales * (float(discount)/100)
netRevenue = totalSales - totalDiscount

totalCost = cost * qty
profit = netRevenue - totalCost
profitPercentage = (profit/totalCost) * 100
profitMargin = round((profit/netRevenue) * 100, 2)

print(f"""

ProductName : {productName}
Sales : {totalSales}
Cost : {totalCost}
Discount : {totalDiscount}
Revenue : {netRevenue}
Profit : {profit}
Profit % : {profitPercentage},
ProfitMargin : {profitMargin}
""")
```

```python
# write a program to check if the given no by user is ODD or EVEN
n = int(input("Enter any no to check if it is an ODD or EVEN: "))

if n%2 == 0:
    print(f"{n} is an EVEN no")
else:
    print(f"{n} is an ODD no")
```

```python
# nested if
marks = 88

if marks>90:
    print("Grade A")
    if marks>99:
        print("Perfect!!")
    elif marks>95:
        print("Brilliant!!")
    else:
        print("Excellent!!")
else:
    print("Grade B")
```

```python
username = "wscube"
password = "1234567"

name = input("Enter user name: ")

if username==name.lower():
    pswd = input("Enter password: ")
    if password==pswd:
        print("Logged-in successfully!!")
    else:
        print("Password incorrect!!")
  
else:
    print("Invalid username!!")
```

```python
name = "Data Analytics"
count = 0

for char in name:
    if char.lower() == "a":
        count += 1
print(f"count of a in {name} is {count}")        
```

```python
name = "Wscube"
count = 0
# write a program to calculate no of vowels (aieou) in the name given

for char in name:
    if char.lower() in "aeiou":
        count += 1

print(f"no of vowels in {name} is {count}")
```

```python
# numerical - range(start, stop+1, step), by default 
# start is 0
# step is 1

for n in range(-1,-10,-1):
    print(n)
```

```python
# write a program to get the sum of all numbers till 50

sum = 0

for n in range(1,51):
    sum += n
    #print(n, sum)
print(sum)    
```

```python
# find the factorial of given no
# 5! = 5*4*3*2*1 = 120

n = int(input("Enter any no for factorial: "))

fact = 1

for i in range(n, 0, -1):
    fact *= i
    print(i, fact)
```

```python
# print a multiplication table of 5

n = int(input("Enter any no: "))

for i in range(1, 11):
    print(f"{n} x {i} = {n*i}")
```

```python
# want to print tables of 1 to 10
# nested for loop

for n in range(70,81):
    for i in range(1, 11):
        print(f"{n} x {i} = {n*i}")
    print("\n")    
```
