```python
# Operators
# Airthmatic -> +, -, *, /, %, **, //
# Comparison -> >, <, >=, <=, ==, != - these all give bool result
# Logical -> and, or, not - these all give bool result
# Assignment -> =, +=, -=, *=...
# Membership -> in, not in - these all give bool result
# Identity operators -> is, is not - these all give bool result

n1 = 5
n2 = 5

print(n1+n2)
print(n1-n2)
print(n1-n2)
print(n1/n2)

print(n1%n2)
print(n2**n1)
print(n1//n2)
```

```python
print(n1>=n2)
print(n1<=n2)
print(n1==n2)
print(n1!=n2)
```

```python
n1 = 5
n2 = 7
n3 = 2

print(not((n1>n2) and (n1>n3)))
```

```python
total_exp = 0

m_amt = float(input("Enter your Movie expense: "))
total_exp += m_amt
s_amt = float(input("Enter your Shopping expense: "))
total_exp += s_amt
r_amt = float(input("Enter rent amount received: "))
total_exp -= r_amt

print(total_exp)
```
