```python
import numpy as np

a = [2,3,5,6]
#b = [4,7,8,2]
b = 4

print(a)
#print(a+b)
#print(a-b)
print(a*b)
```

```python
x = np.array(a)
y = np.array(b)
print(x)
print(y)
print(x+y)
print(x-y)
print(x*y)
print(x/y)
```

```python
import statistics as stats
price = [300, 500, 200, 800, 500]
qty = [6, 7, 4, 8, 4]

x = np.array(price)
y = np.array(qty)

print(np.sum(x))
print(np.sum(y))
print(np.cumsum(x))
print(np.cumprod(y))
print(np.ptp(x))
print(np.ptp(y))
print(np.mean(x))
print(np.median(x))
print(stats.mode(x))
print(stats.mode(y))
print(np.std(x))
print(np.var(x))
```

```python
# 1d
a = [1,
     3,
     4,
     2,
     6,
     5,
     7,
     8,
     4,
     5,
     4,
     2]

x = np.array(a)
print(x)
print(np.ndim(x))
print(np.shape(x))
print(np.size(x))
print(len(x))
```

```python
# 2d
a = [
     [1,3],
     [4,2],
     [6,5],
     [7,8],
     [4,5],
     [4,2]
    ]

x = np.array(a)
print(x)
print(np.sum(x, axis=1))
print(np.ndim(x))
print(np.shape(x))
print(np.size(x))
print(len(x))
```

```python
print(np.reshape(x, (4,3)))
```

```python
# 3d
a = [
        [
            [3,4,5],
            [4,6,7],
            [7,8,9]
        ],
        [
            [2,3,1],
            [3,6,7],
            [5,6,9]
        ],
        [
            [9,8,1],
            [7,6,2],
            [2,5,3]
        ],
        [
            [9,7,5],
            [5,6,4],
            [2,1,3]
        ]
]

x = np.array(a)
print(x)
print(np.ndim(x))
print(np.shape(x))
print(np.size(x))
print(len(x))
```

```python
# indexing 
a = [
     [1,3],
     [4,2],
     [6,5],
     [7,8],
     [4,5],
     [4,2]
    ]

x = np.array(a)
print(x[3][-1])
```

```python
print(x[4][0])
print(x[-2][-2])
```

```python
# slicing
a = [
     [1,3],
     [4,2],
     [6,5],
     [7,8],
     [4,5],
     [4,2]
    ]

x = np.array(a)
print(x[-3])
print(x[-3:])
```
