```python
f = open("demo.txt", 'w')
f.write({
    "name" : "Raman",
    "age" : 16
})
f.close()

# with open("demo.txt", 'w') as f:
    
```

```python
f = open("demo.txt", 'a')
f.write("\nHow are you?")
f.close()
```

```python
f = open("demo.txt", 'r')
data = f.readlines()
print(data)
f.close()
```

```python
import operations as op

print(op.add(5,6))
print(op.sub(5,6))
print(op.prod(5,6))
print(op.divide(5,6))
```

```python
import pickle as p

f = open("demo.txt", 'wb')
p.dump({
    "name" : "Raman",
    "age" : 16
},f)
f.close()
```

```python
with open("demo.txt", 'ab') as f:
    p.dump({
        "name" : "Raghav",
        "age" : 15
    }, f)
```

```python
f = open("demo.txt", 'rb')
data = p.load(f)
data2 = p.load(f)
print(data, data2)
f.close()
```

```python
# module vs package vs library
# module - a predefined reusable set of 
#          instruction provided in one single file of python
# pickle, math, random, datetime
# package - a collection of similar modules or a 
#           directory holding multiple modules
# library - it is a collection of reusable set of instructions
#           created for a certain task in python
# numpy, pandas, matplotlib, seaborn
```
