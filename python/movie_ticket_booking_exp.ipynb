```python
# movie ticket booking system
# loops
# nested collections
# logic building

movies_db = {
    "Hanuman Ansh" : {
        "seats" : 10,
        "price" : 500
    },
    "Bahubali" : {
        "seats" : 5,
        "price" : 400
    },
    "RRR" : {
        "seats" : 15,
        "price" : 200
    },
    "Ramayana" : {
        "seats" : 12,
        "price" : 500
    },
    "Kantara" : {
        "seats" : 7,
        "price" : 700
    }
}

# for k,v in movies_db.items():
#     print(k, ":", v)

while True:

    print(f"""
            Welcome to Box Office
    Press 1 for Displaying Movies        
    Press 2 for Displaying Price 
    Press 3 for Displaying available Seats
    Press 4 for Booking Tickets 
    Press 5 for Cancelling Tickets
    Press q to Quit
    """)

    choice = input("Enter your choice: ")

    if choice in ("q", "Q", "Quit", "quit", "QUIT"):
        break

    if choice=="1":
        for movie in movies_db.keys():
            print(movie)
    elif choice=="2":
        for k,v in movies_db.items():
            print(k, ": Rs", v["price"])
    elif choice=="3":
        for k,v in movies_db.items():
            print(k, ":", v["seats"])
    elif choice=="4":
        for k,v in movies_db.items():
            print(k, ":", v)

        movie = input("Enter movie name:")
        movie_name = movie.title()
        if movie_name in movies_db.keys():
            noOfSeats = int(input("Enter no of seats: "))
            if movies_db[movie_name]["seats"] >= noOfSeats:
                movies_db[movie_name]["seats"] -= noOfSeats
                print("-"*50)
                totalPrice = movies_db[movie_name]["price"] * noOfSeats
                tax = totalPrice * 0.1
                print(f"{noOfSeats} tickets booked successfully!!")
                print(f"Total Amout : {totalPrice}")
                print(f"TAX : {tax}")
                print(f"Total Price : {totalPrice + tax}")
                print("-"*50)
            else:
                print(f"Insufficients seats, out of {noOfSeats}, we have only {movies_db[movie_name]['seats']} available!!")    
            
        else:
            print("Movie not on the Box Office!!")
    elif choice == "5":
        movie = input("Enter movie name:")
        movie_name = movie.title()

        if movie_name in movies_db.keys():
            noOfSeats = int(input("Enter no of seats: "))
            movies_db[movie_name]["seats"] += noOfSeats
            refunded_price = movies_db[movie_name]["price"]*0.5
            total_price = refunded_price*noOfSeats
            print("-"*50)
            print(f"{noOfSeats} tickets have been cancelled successfully!!")
            print(f"Amount Refunded: {total_price}")
            print("-"*50)
            
        else:
            print("Movie not on the Box Office!!")
    else:
        print("Invalid choice, choose a correct option from the menu!!")



        
        
```
